````python
# Intentionally broken code for /fix demo
def calculate_average(numbers):
    total = 0
    for num in numbers
        total += num
    return total / len(numbers)

# Missing colon and wrong indentation
result = calculate_average([1, 2, 3, 4, 5]
print(f"Average: {result}")
````

## Demo Input/Output Cells for Interpolation Examples

The following cells provide actual input, output, and error examples that will be referenced in the interpolation demos below.


````python
# Cell to be referenced by {In[23]} - Data analysis example
import pandas as pd
import matplotlib.pyplot as plt

# Create sample dataset
data = {'product': ['A', 'B', 'C', 'D'], 
        'sales': [100, 150, 200, 120],
        'profit': [20, 30, 50, 25]}
df = pd.DataFrame(data)

# Calculate profit margin
df['profit_margin'] = (df['profit'] / df['sales'] * 100).round(2)
print("Product Performance Analysis:")
print(df)
````

````python
# Cell to be referenced by {Out[24]} - Analysis results
best_product = df.loc[df['sales'].idxmax(), 'product']
total_sales = df['sales'].sum()
avg_margin = df['profit_margin'].mean()

results = f"""
SALES ANALYSIS RESULTS:
- Best performing product: {best_product}
- Total sales across all products: ${total_sales}
- Average profit margin: {avg_margin:.1f}%
- Recommendation: Focus marketing on product {best_product}
"""
print(results)
results  # This will be the output that can be referenced
````

````python
# Cell to be referenced by {Err[25]} - Intentional error for demo
def calculate_roi(profit, investment):
    return (profit / investment) * 100

# This will cause a ZeroDivisionError
roi = calculate_roi(100, 0)  # Division by zero!
print(f"ROI: {roi}%")
````
